Sure! Here's a detailed **commentary-only** outline of how to approach solving the 1D inviscid Burgers' equation numerically using both the **flux-conservative** and **non flux-conservative** upwind schemes. This includes setup, methodology, and what to expect when changing resolution.

---

### 💡 Problem Overview:

You're solving the 1D nonlinear **inviscid Burgers' equation**:

$$
\frac{\partial u}{\partial t} + u \frac{\partial u}{\partial x} = 0
$$

with the **initial condition**:

$$
u(x, 0) = 10 e^{-(x - x_0)^2}, \quad x_0 = 5
$$

on the domain $x \in [0, 10]$, using **periodic boundary conditions**. Time evolution runs until $t = 0.5$, and you’re comparing two schemes: flux-conservative and non-conservative upwind.

---

### 🧱 Grid and Time Setup:

* **Grid**: Use at least 101 points → spacing $\Delta x = \frac{10}{J - 1}$
* **Courant factor**: $\text{cf} = 0.5$
* **Time step**: $\Delta t = \text{cf} \cdot \frac{\Delta x}{u_{\max}}$, where $u_{\max}$ is initially 10
* **Termination time**: Run until $t = 0.5$

---

### 🚦 Non-Conservative Upwind Scheme:

* Discretizes the equation directly in its non-conservative form.
* Use **upwind differencing** depending on the sign of $u$ at each point (since the advection speed is $u$ itself).
* This approach may introduce **numerical dissipation** and may **not capture shock formation** accurately over time.
* Tends to smear discontinuities, especially at lower resolutions.

---

### 🌊 Flux-Conservative Upwind Scheme:

* Rewrite the equation in **conservative form**:

$$
\frac{\partial u}{\partial t} + \frac{\partial}{\partial x} \left( \frac{1}{2} u^2 \right) = 0
$$

* Then discretize the **flux term** $f(u) = \frac{1}{2} u^2$ using upwind logic, again based on the sign of local velocities.
* This version preserves conserved quantities better and is **more robust in capturing shock formation**.
* You'll likely see steeper fronts or even developing discontinuities compared to the smeared-out solution from the non-conservative scheme.

---

### 🧪 Results and Observations:

* **At t = 0.5**, compare both schemes:

  * The **non-conservative** version will have a smoother profile.
  * The **flux-conservative** version will likely develop a sharper gradient or even a shock.
* **As you increase resolution**:

  * Both schemes converge to the true solution more closely.
  * **Conservative scheme** becomes more accurate in capturing steep gradients or shocks.
  * **Non-conservative scheme** still diffuses these features more, though less so at higher resolutions.

---

### 📈 Plotting Recommendations:

  * $t = 0.5$

### 🔍 Final Remarks:

* Burgers’ equation naturally develops shocks from smooth initial data.
* **Conservative schemes** are critical for correctly capturing these shocks.
* **Non-conservative schemes** can violate physical conservation laws, especially at low resolutions.
* Choosing the correct scheme matters **more** when nonlinearities dominate and shock formation is expected.

Great question — you're observing a **very important effect** in numerical solutions of nonlinear PDEs like the inviscid Burgers' equation.

Let’s break it down:

---

### 🔁 1. **Why does the peak go down over time?**

In the **exact solution** of the inviscid Burgers' equation, the **peak does not go down** — the total area under $u(x, t)$ (i.e. the conserved quantity) should stay the same in time (assuming the domain is periodic and no mass leaves it). However, in numerical simulations:

#### ✅ In the **flux-conservative** scheme:

* Conservation laws (like mass or momentum) are better preserved.
* You may still see **some peak reduction**, especially at lower resolution, due to numerical **dissipation** and the way the shock is represented.
* But overall, the area under the curve (integral of $u$) is more accurate, and the peak doesn’t drop significantly unless the shock is under-resolved.

#### ❌ In the **non-flux-conservative** scheme:

* You're not solving the PDE in conservative form, so **conserved quantities aren't guaranteed to stay constant**.
* As a result, numerical artifacts can cause "mass loss" — the **area under the curve decreases**, and the **peak flattens**.
* The scheme acts like it’s leaking energy or fluid out of the system.

---

### 🔍 2. **Why is this more prominent in the non-conservative scheme?**

Because the equation is nonlinear, small numerical errors **compound and feed back into the solution**.

Specifically:

* In the **non-conservative form**, you're computing $u \partial_x u$ directly. This isn't equivalent to computing the derivative of a flux, especially when discretized.
* Any asymmetry or discretization error here **does not account for the balance of fluxes** across the grid, so over time, the method can:

  * Undershoot peak values (drop the max)
  * Introduce artificial viscosity (smoothing out gradients)
  * Fail to form or preserve shock-like features
* This results in a **smeared and flattened** profile — the peak drops more than it should, and the solution doesn't look as physically accurate.

---

### 🧠 Bonus: Why flux-conservative schemes help

In the conservative form:

$$
\frac{\partial u}{\partial t} + \frac{\partial}{\partial x} \left( \frac{1}{2} u^2 \right) = 0
$$

you're ensuring that **the net flux in and out of each cell** balances the rate of change of $u$ in that cell. That means:

* If a wave (or mass) leaves one cell, it must enter the neighboring one.
* So **total "mass" is conserved** up to machine error, even across shocks.

This is why **shock capturing** methods (e.g., Godunov-type schemes) are based on conservative forms.

---

### 🔚 Summary:

* The **peak drops** mainly due to **numerical dissipation**.
* It’s **worse in the non-conservative form** because conservation laws are violated numerically.
* Conservative schemes **preserve total quantities** better and thus maintain sharper features, even as resolution improves.

Excellent follow-up — both **grid resolution** and the **Courant factor (CF)** affect your numerical solution, but **in different ways**.

---

### 🎯 1. **Changing Resolution (Number of Grid Points $J$)**

* **Directly improves spatial accuracy.**
* Smaller $\Delta x = \frac{L}{J - 1}$ means:

  * Sharper gradients and shocks are better resolved.
  * Less numerical dissipation (especially in conservative schemes).
* Crucial when studying **shock formation** or **steep wave fronts** like in Burgers’ equation.
* For both flux and non-flux conservative schemes, **higher resolution reduces peak drop**, but it helps **conservative schemes more**, because the shock becomes better defined.

📌 So: **Increasing resolution** is the primary way to get more accurate results.

---

### ⏱️ 2. **Changing the Courant Factor (CF = $\frac{u_{\max} \Delta t}{\Delta x}$)**

* Affects **time step size**, not spatial resolution.
* Lower CF → smaller time step → **more accurate time integration**, but slower simulation.
* Higher CF (closer to 1) pushes the **stability limit** of the explicit upwind scheme.

  * Might still be stable, but can introduce more **numerical error**, especially in nonlinear problems.
* For **CF too high (>1)**, the method becomes **unstable** (blows up).
* For **CF too low**, the solution is stable and accurate, but inefficient (takes many steps).

📌 So: CF impacts **temporal accuracy and stability**, but not the inherent spatial resolution.

---

### 🤝 Key Interaction Between the Two:

* If you **increase resolution**, $\Delta x$ gets smaller, so to **keep CF fixed**, you must also reduce $\Delta t$.
* Holding CF constant as you refine the grid ensures that both time and space errors are balanced (i.e., you're not under-resolving in time while over-resolving in space).

---

### 🔑 Bottom Line:

| Parameter           | What It Affects                                            | Impact on Solution            |
| ------------------- | ---------------------------------------------------------- | ----------------------------- |
| Grid resolution $J$ | **Spatial accuracy**                                       | Higher = better shock capture |
| Courant factor (CF) | **Time step & stability**                                  | Lower = safer, but slower     |
| Combined change     | Must scale $\Delta t$ with $\Delta x$ to maintain accuracy |                               |

If you’re doing convergence studies, **always increase both resolution and decrease $\Delta t$** proportionally (by keeping CF fixed) to avoid mixing spatial and temporal errors.
